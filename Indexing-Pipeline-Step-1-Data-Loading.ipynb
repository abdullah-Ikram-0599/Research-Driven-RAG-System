### Loading Data from Wikipedia

```python
from langchain_community.document_loaders import WikipediaLoader
import wikipedia

wikipedia.set_user_agent( # Setting a custom user agent for Wikipedia requests
    "my-rag-project/1.0"
)

loader = WikipediaLoader(
   query = "Retrieval-augmented generation", # The query to search for on Wikipedia
   load_max_docs=1,  # The maximum number of documents to load from Wikipedia
   doc_content_chars_max=1000000 # The maximum number of characters to load from each document
) 

docs_wiki = loader.load() # Loading the documents from Wikipedia

# print(docs_wiki[0].page_content) # Printing the content of the first document
```

### Loading Data from PDF stored locally
